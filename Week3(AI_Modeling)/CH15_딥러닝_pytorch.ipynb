### 2) ⚡ Autograd 마법 체험 — 미분이 저절로 된다

- 함수 y = x²
- 미분 y = 2x
- x=3에서 접선의 기울기는 6
- PyTorch는 `.backward()` **한 줄**로 계산합니다.

<details>
<summary> ---------- 💡 예시 코드 보기 ---------- </summary>

```python
# requires_grad=True: "이 숫자에 대한 미분을 추적해줘"라는 표시
x = torch.tensor(3.0, requires_grad=True)

y = x ** 2          # y = x²
y.backward()        # ← 마법의 한 줄: y를 x로 미분

print(x.grad)       # 6.0 — 2x에 x=3을 넣은 값이 저절로 나옴
```

</details>

````python
# requires_grad=True: 이 값에 대한 미분 계산을 추적하겠다는 설정




````

<details>
<summary> ---------- 💡 예시 코드 보기 ---------- </summary>

```python
# 더 복잡해도 똑같이 한 줄 — y = x³ + 2x, x=2에서 미분하면 3x² + 2 = 14

y = x ** 3 + 2 * x
y.backward()

```

</details>

````python
# 더 복잡해도 똑같이 한 줄
# y = x³ + 2x
# 미분하면 3x² + 2 = 14
# x=2에서의 기울기는 14




````

> 💡 **이게 왜 마법인가요?** 딥러닝의 학습 = "loss를 줄이는 방향(기울기) 찾기 = 미분"인데, 실제 모델은 파라미터가 수십만~수십억 개입니다. 그 미분을 전부 자동으로 해주는 것이 Autograd이고, 어제 Keras의 `fit()`이 내부에서 조용히 하던 일이 바로 이것입니다. 오늘 2교시에 우리가 직접 쓸 `loss.backward()` 한 줄의 정체이기도 합니다.

### 2) 데이터 공급 라인 — Dataset과 DataLoader

수만 장을 통째로 메모리에 올려 학습시키는 건 실무에선 불가능합니다. PyTorch는 "데이터를 배치(묶음) 단위로 공급하는 라인"을 두 부품으로 만듭니다:

| 부품 | 역할 | 우리가 할 일 |
|---|---|---|
| `Dataset` | 데이터 창고 — "몇 개 있고, i번째는 뭔지"만 답하면 됨 | `__len__`, `__getitem__` 두 메서드 구현 |
| `DataLoader` | 컨베이어 벨트 — 창고에서 batch_size만큼 꺼내 섞어서 공급 | 만들어진 걸 쓰기만 |

이 패턴을 한 번 익히면 **회사의 어떤 데이터든** 딥러닝에 태울 수 있습니다 — 그래서 이건 직접 만들어봅니다.

class FashionDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)     # 특성은 float32
        self.y = torch.tensor(y, dtype=torch.long)          # 정수 라벨은 long

    def __len__(self):
        return len(self.X)                 # 창고에 몇 개?

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]     # i번째 (입력, 정답)은?

train_dataset = FashionDataset(X_train, y_train)
test_dataset = FashionDataset(X_test, y_test)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)    # 훈련: 섞어서 32개씩

### 4) 학습 루프 — 오늘의 심장, 직접 씁니다

```

backward() → 오답 분석하기 (기울기 계산)
step() → 가중치 수정하기
```

def train_loop(use_zero_grad, epochs=5):

    m = FashionNet()

    losses = []

### 규칙 정리 — 부수면 안 되는 두 스위치

실험으로 확인한 것 외에, 외워야 할 안전 규칙 두 가지:

| 규칙 | 안 지키면 |
|---|---|
| 학습 전 `model.train()` / 평가 전 `model.eval()` | Dropout·BatchNorm 같은 부품이 잘못된 모드로 동작 (지금 모델엔 없지만, 실무 모델엔 거의 항상 있음) |
| 평가는 `with torch.no_grad():` 안에서 | 시험 중에도 기울기를 추적해 메모리 낭비 + 느려짐 |

## 📋 PyTorch 학습 루프 치트시트

```
[매 배치]  zero_grad → model(X) → loss → backward → step
[학습 전]  model.train()
[평가 전]  model.eval() + torch.no_grad()
[디버깅]   이상하면 zero_grad부터 확인
```

---

## ⚠️ 자주 하는 실수

> ⚠️ zero_grad() 누락 — 에러 없이 조용히 망가지는 실무 버그 1위 (실험 A).

> ⚠️ 평가 루프에서 no_grad를 빼먹어도 결과는 같게 나옵니다 — 그래서 습관이 안 잡히는데, 데이터가 커지면 메모리 부족으로 터집니다.

### 2) device 3종 세트 — GPU가 있든 없든 돌아가는 코드

실무 PyTorch 코드는 GPU 유무와 상관없이 돌아가야 합니다. 외울 것은 세 가지뿐:

| 패턴 | 역할 |
|---|---|
| `device = torch.device("cuda" if ... else "cpu")` | 장치 자동 선택 |
| `model.to(device)` | 모델을 장치로 |
| `batch_X.to(device), batch_y.to(device)` | 매 배치 데이터를 장치로 (루프 안에서) |

| | Keras | PyTorch |
|---|---|---|
| 강점 | 빠른 프로토타입, 간결함 | 내부 제어, 새 구조 실험 |
| 주 사용처 | 서비스 프로토타입, 교육 | 연구, 최신 모델 구현체 |
| 실무 결론 | **둘 다 읽고 쓸 줄 알면** 어떤 회사·논문 코드도 두렵지 않음 — 여러분은 이제 이틀 만에 그 상태입니다 |

**4일의 마무리**: 회귀·분류(트리) → 비지도·튜닝 → Keras → PyTorch까지, "데이터 → 모델 → 평가 → 개선"이라는 하나의 사이클을 네 가지 도구로 반복했습니다. 도구는 바뀌어도 사이클은 같았죠 — 그 사이클이 여러분이 이번 주에 얻은 진짜 자산입니다.

**YOLO(You Only Look Once)**는 객체 탐지의 대표 모델 계열로, 사진을 "한 번만 보고" 모든 객체의 위치와 종류를 동시에 찾아냅니다. 올해 1월 공개된 최신 버전 **YOLO26**은 GPU 없이 CPU에서도 빠르게 돌아가도록 설계되어, 지금 여러분의 노트북에서도 최신 비전 AI를 직접 돌려볼 수 있습니다.

> 💡 원리(CNN 구조, 아키텍처 계보)는 다음 개념 교안에서 자세히 다룹니다 — 오늘은 "최신 모델을 내 손으로 돌려보는 경험"에 집중합니다.

result = results[0]

# 무엇이 몇 개 탐지되었는지 집계
from collections import Counter

detected = [result.names[int(c)] for c in result.boxes.cls]
print("탐지된 객체:", dict(Counter(detected)))
print(f"→ 현재 화면 인원: {detected.count('person')}명")
