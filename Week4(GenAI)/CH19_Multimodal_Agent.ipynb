### 3) 🎨 오늘의 하이라이트 ① — 내 목소리를 내 모델이 받아쓴다

아래 셀을 실행하면 녹음 버튼이 나타납니다 (Colab 전용). **영어로 여행 문의 한 문장**을 말하고 멈추면, 방금 그 목소리를 Whisper가 받아씁니다.

> 💡 Colab이 아닌 환경에서는 녹음이 동작하지 않습니다 — 그 경우 이 셀을 건너뛰고 배포된 `travel_inquiry.wav`로 진행하면 됩니다. 소형(tiny) 모델이라 한국어 인식률은 낮으니 영어로 시도하는 것을 권장합니다 — 더 큰 whisper 모델은 한국어도 잘합니다(자습 과제).

````python
# 내 목소리 녹음하기 (Colab 전용 — 다른 환경에서는 건너뛰세요)

from IPython.display import Javascript
from google.colab import output
from base64 import b64decode

RECORD_JS = """
const sleep = time => new Promise(resolve => setTimeout(resolve, time))
const b2text = blob => new Promise(resolve => {
  const reader = new FileReader()
  reader.onloadend = e => resolve(e.srcElement.result)
  reader.readAsDataURL(blob)
})
var record = time => new Promise(async resolve => {
  stream = await navigator.mediaDevices.getUserMedia({ audio: true })
  recorder = new MediaRecorder(stream)
  chunks = []
  recorder.ondataavailable = e => chunks.push(e.data)
  recorder.start()
  await sleep(time)
  recorder.onstop = async () => {
    blob = new Blob(chunks)
    text = await b2text(blob)
    resolve(text)
  }
  recorder.stop()
})
"""

def record_audio(seconds=10, filename="my_voice.webm"):
    print(f"🎙️ {seconds}초 동안 녹음합니다 — 영어로 여행 문의 한 문장을 말해보세요!")
    from google.colab.output import eval_js
    eval_js(RECORD_JS)
    data = eval_js(f"record({seconds * 1000})")
    binary = b64decode(data.split(",")[1])
    with open(filename, "wb") as f:
        f.write(binary)
    print("녹음 완료 →", filename)
    return filename

my_audio = record_audio(seconds=10)
````

    plt.subplot(1, 5, i + 1)
    plt.imshow(frame)
    plt.axis("off")
    plt.title(f"노이즈 {int(r * 100)}%")
plt.suptitle("Diffusion의 아이디어 — 잡음에서 이미지를 '조각'해내기", y=1.05)

def route_v01(question):
    if "날씨" in question:
        return "get_weather"
    elif "환율" in question:
        return "get_exchange_rate"
    elif any(op in question for op in ["+", "-", "*", "/", "곱하기", "더하기"]):

### 2) MCP의 구조 — 세 명의 등장인물

| 역할 | 하는 일 | 컨시어지 비유 |
|---|---|---|
| **Host** | 사용자와 대화하는 AI 앱 | AI 여행 컨시어지 앱 |
| **Client** | Host ↔ Server 사이의 표준 통신 담당 | 규격에 맞춰 주문서를 전달하는 직원 |
| **Server** | 실제 도구를 표준 규격으로 제공 | 날씨 서버, 캘린더 서버, 호텔 서버... |

핵심 약속은 두 가지입니다: 모든 서버는 **① "내가 가진 도구 목록"을 보여줄 수 있어야 하고(메뉴판), ② 표준 방식으로 도구를 실행해줄 수 있어야 한다(주문).** 이 약속만 지키면 어떤 도구든 어떤 AI 앱에든 꽂을 수 있습니다.

class ConciergeMCPServer:
    def __init__(self, name):
        self.name = name
        self.tools = {}

        return {tool_name: t["description"] for tool_name, t in self.tools.items()}

    def call_tool(self, tool_name, *args):

        if tool_name not in self.tools:

    tool, score = route_v10(question)
    if score >= TOOL_THRESHOLD:
        arg = extract_argument(question, tool)
        result = TOOLS[tool](arg)

def build_router_from_servers(servers):
    specs, funcs = {}, {}
    for server in servers:
        for tool_name, desc in server.list_tools().items():
            specs[tool_name] = desc
            funcs[tool_name] = server.tools[tool_name]["func"]
    names = list(specs.keys())
    embs = embedding_model.encode(list(specs.values()))
    def route(question):
        q = embedding_model.encode(question)
        sims = [cosine_similarity(q, e) for e in embs]
        return names[int(np.argmax(sims))], max(sims)
    return route, funcs

- 멀티모달 = 다른 형태의 입력(사진·소리)을 **텍스트 세계로 데려오는 것**에서 시작
- Agent의 핵심은 도구가 아니라 **"어떤 도구를 쓸지 고르는 판단"** — 그리고 그 판단은 의미(임베딩) 매칭
- MCP = 도구 연결의 USB 규격 — **메뉴판 공개 + 표준 실행**
- 3주간 배운 프롬프트(CH17) → RAG(CH18) → 멀티모달·에이전트(CH19)가 하나의 서비스 구조로 연결
